## Step 2: missing values and Cleaning Data

```python
df.isnull().sum()
```

```python
df['director']=df['director'].fillna('unkown')
```

```python
df['country']=df['country'].fillna('unkown')
```

```python
df['cast']=df['cast'].fillna('no')
```

```python
df['date_added']=df['date_added'].fillna('no')
```

```python
df.dropna(subset=['rating'], inplace=True)
```

```python
df['duration']=df['duration'].fillna('no data')
```

## Step 3: Movies and TV shows using (Bar Chart)

```python
df['type'].value_counts().reset_index(name='count')
```

```python
content_type=df['type'].value_counts()

plt.bar(content_type.index, content_type.values,color=['#1f77b4', '#ff7f0e'],edgecolor='black',label=content_type)

plt.title('Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')
plt.grid(alpha=0.2)
plt.legend()
plt.show()
```

## Step 4: Top 10 Countries (Bar Chart)

```python
top_countries=df['country'].value_counts().head(10)
```

```python
top_countries.reset_index(name='count')
```

```python
plt.figure(figsize=(12,5))

colors_bar = ['#3498DB', '#F39C12', '#E74C3C', '#9B59B6','#1ABC9C','#3498DB', '#F39C12', '#E74C3C', '#9B59B6','#1ABC9C']
plt.bar(top_countries.index, top_countries.values,color=colors_bar,label=top_countries,edgecolor='black')
plt.xticks(rotation=45)

plt.title('Top 10 Countries with Netflix Content')
plt.xlabel('Countries')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.4)
plt.legend()
plt.show()
```
